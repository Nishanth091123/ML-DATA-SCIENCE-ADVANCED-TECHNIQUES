# Wine Classification | Data Science and Machine Learning
**Dataset:** UCI Wine, available locally through scikit-learn. 178 samples, 13 numerical predictors, three classes.

**Dataset source:** https://archive.ics.uci.edu/dataset/109/wine

```python
from pathlib import Path
import pandas as pd
from sklearn.datasets import load_wine
base = Path.cwd()
wine = load_wine(as_frame=True)
df = wine.frame.rename(columns={'target':'class'})
print('Dataset shape:',df.shape)
print('Feature names:',list(df.drop(columns='class').columns))
print('Target classes:',wine.target_names)
print(df.head().to_string(index=False))
print('Class distribution:')
print(df['class'].value_counts().sort_index())
df.to_csv(base/'wine_raw.csv',index=False)
print('Saved:',base/'wine_raw.csv')
```

**Observation:** There are 178 samples, 13 numerical measurements and a class label. The class is an identifier (0, 1, 2), not a continuous quantity.
